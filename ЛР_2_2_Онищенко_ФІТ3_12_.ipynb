## Завдання 1. Титанік

Датасет завантажую з Kaggle через kagglehub.

```python
import kagglehub
import os

# скачую датасет з Kaggle
path = kagglehub.dataset_download("mahmoudshogaa/titanic-dataset")

print("Path to dataset files:", path)
```

```python
df = pd.read_csv(os.path.join(path, "titanic.csv"))

df
```

Що означають колонки:
- **PassengerId**: номер пасажира
- **Survived**: вижив (1) чи ні (0)
- **Pclass**: клас квитка (1 найвищий, 3 найнижчий)
- **Name**: ім'я
- **Sex**: стать

```python
df.duplicated().sum()
```

4. Пропущені значення

```python
# пропуски по колонках
df.isnull().sum()
```

Пропуски є в Age, Cabin і Embarked. Вік заповнила медіанним значенням, а колонки Cabin і Embarked видалила, бо далі вони не знадобляться.

```python
# заповнюю пропуски у віці медіаною
df['Age'] = df['Age'].fillna(df['Age'].median())

# видаляю непотрібні колонки
df.drop(columns=['Cabin', 'Embarked'], inplace=True)
```

5. Перевіряю ще раз

```python
df.isnull().sum()
```

6. Залишаю тільки потрібні колонки: Survived, Pclass, Sex, Age, Fare

```python
# залишаю тільки потрібні колонки
df = df[['Survived', 'Pclass', 'Sex', 'Age', 'Fare']].copy()

df.head()
```

7. Стать переводжу в 0 і 1. Спочатку дивлюсь, які є значення:

```python
df['Sex'].value_counts()
```

Значень лише два: male і female. Кодую male = 0, female = 1.

```python
# стать у числа
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})

print(f'Data types of the dataset:\n{df.dtypes}')
```

8. Описова статистика

```python
df.describe()
```

9. Пропусків більше немає

```python
df.isnull().sum()
```

10. Перші 5 рядків

```python
df.head()
```

11. Останні 5 рядків

```python
df.tail()
```

12. Виживання за статтю

```python
survival_by_sex = df.groupby('Sex')['Survived'].mean() * 100

print(f'Share of survived males: {survival_by_sex[0]:.2f}%')
print(f'Share of survived females: {survival_by_sex[1]:.2f}%')
```

```python
print(f'Correlation between sex and survival: {df["Sex"].corr(df["Survived"])}')
```

Різниця дуже велика: вижило 74.2% жінок і тільки 18.9% чоловіків. Кореляція статі з виживанням 0.54.

13. Виживання за класом

```python
# частка тих, хто вижив, у кожному класі
class_surv = df.groupby('Pclass')['Survived'].mean() * 100

for cls, rate in class_surv.items():
    print(f'Share of survived class {cls} passengers: {rate:.2f}%')
```

Найвища частка тих, хто вижив, у першому класі (63%), найнижча у третьому (24%).

14. Середній вік тих, хто вижив і не вижив

```python
survived_mean_age = df[df['Survived'] == 1]['Age'].mean()
not_survived_mean_age = df[df['Survived'] == 0]['Age'].mean()

print(f'Mean age of survived passengers: {survived_mean_age:.2f}')
print(f'Mean age of not survived passengers: {not_survived_mean_age:.2f}')
```

28.3 і 30.0 років, різниця невелика. Тобто за середнім віком вплив на виживання майже не помітний.

15. Виживання за рівнем тарифу

```python
# ділю пасажирів на 5 груп за ціною квитка
df['Fare_Category'] = pd.qcut(df['Fare'], 5, labels=False)

# частка тих, хто вижив, у кожній групі
for i in range(5):
    surv_rate = df[(df['Fare_Category'] == i) & (df['Survived'] == 1)].shape[0] / df[df['Fare_Category'] == i].shape[0]
    print(f'Share of survived passengers in bucket {i}: {surv_rate * 100:.2f}%')
```

Що дорожчий квиток, то більше шансів вижити: у двох найдешевших групах вижило близько 20-22%, у найдорожчій 64%.

16. Тариф за класами

```python
# медіанна ціна квитка для кожного класу
for i in range(1, 4):
    median_fare = df[df['Pclass'] == i]['Fare'].median()
    print(f'Median fare for class {i}: {median_fare:.2f}')
```

Різниця суттєва. Перший клас приблизно в 4 рази дорожчий за другий, а другий майже в 2 рази дорожчий за третій.

17. Вік і виживання на графіку

```python
# об'єдную дані в один датафрейм
df_survived = df[df['Survived'] == 1][['Age']]
df_survived['Status'] = 'Survived'

df_not_survived = df[df['Survived'] == 0][['Age']]
df_not_survived['Status'] = 'Not Survived'

df_combined = pd.concat([df_survived, df_not_survived])

# гістограма віку
sns.histplot(data=df_combined, x='Age', hue='Status', bins=25)
plt.title('Розподіл віку пасажирів за виживанням')
plt.show()
```

Що видно з гістограми:
- найбільше пасажирів у віці 20-30 років, і серед них набагато більше тих, хто не вижив;
- серед дітей до 20 років частка тих, хто вижив, вища, мабуть дітей рятували першими;
- у 30-50 років тих, хто не вижив, теж більше, але різниця менша;
- після 50 пасажирів мало, і вижило приблизно стільки ж, скільки загинуло;
- у групі 70+ людей зовсім мало.

Тобто хоча середній вік майже однаковий, по групах видно, що діти мали більше шансів, а молоді дорослі 20-30 років загинули найбільше.

Додатково подивилась виживання за класом і статтю одночасно:

```python
# виживання за класом і статтю

for i in range(1, 4):
    for j in range(2):
        surv_rate = df[(df['Pclass'] == i) & (df['Sex'] == j) & (df['Survived'] == 1)].shape[0] / df[(df['Pclass'] == i) & (df['Sex'] == j)].shape[0]
        print(f'Share of survived {"male  " if j == 0 else "female"} passengers in class {i}: {surv_rate * 100:.2f}%')
```

І матриця кореляції ознак:

```python
mtx = df.drop('Survived', axis=1).corr(numeric_only=True).abs()

# теплова карта кореляції
fig, ax = plt.subplots(figsize=(6, 6))

sns.heatmap(
    mtx,
    cmap='Blues',
    annot=True,
    fmt=".2f",
    linewidths=0.5,
    mask=np.triu(np.ones_like(mtx, dtype=bool)),  # верхній трикутник не показую
    square=True,
    cbar=False,
    ax=ax
)

plt.title("Матриця попарної кореляції ознак", fontsize=14)
plt.tight_layout()
plt.show()
```

**Висновок.** Найсильніше на виживання вплинули стать і клас. З жінок першого і другого класу вижили майже всі (97% і 92%), а з чоловіків другого і третього класу лише 16% і 14%. Дорожчий квиток теж підвищував шанси. Клас найбільше пов'язаний з ціною квитка (0.55, а з групою тарифу 0.71).
